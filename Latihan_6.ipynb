1. Buat program yang meminta sebuah angka, lalu tampilkan "Kelipatan 5" jika angka tersebut habis dibagi 5

```python
angka = int(input("Masukkan sebuah angka: "))

if angka % 5 == 0 and angka != 0:
    print("Kelipatan 5")
elif angka == 0:
    print("Angka adalah nol")
else:
    print("Bukan kelipatan 5")
```

2. Menentukan tahun kabisat dengan menggunakan kondisi
- habis dibagi 400 -> kabisat
- habis dibagi 4 dan tidak habis dibagi 100 -> kabisat

```python
tahun = int(input("Masukkan tahun: "))

if tahun % 400 == 0:
    print(f"Tahun {tahun} adalah Tahun Kabisat (Habis dibagi 400).")
elif tahun % 4 == 0 and tahun % 100 != 0:
    print(f"Tahun {tahun} adalah Tahun Kabisat (Habis dibagi 4 dan tidak habis dibagi 100).")
else:
    print(f"Tahun {tahun} bukan Tahun Kabisat.")
```
