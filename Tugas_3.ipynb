1. Berdasarkan dokumentasi resmi Python, setiap operator memiliki urutan prioritas (operator precedence) yang menentukan operator mana yang dievaluasi (diselesaikan) lebih dahulu dalam sebuah ekspresi.
Operator dengan prioritas lebih tinggi akan dikerjakan sebelum operator dengan prioritas lebih rendah. Pada Python, operator perkalian (*) memiliki prioritas lebih tinggi daripada operator penjumlahan (+).


```python
10 + 5 * 2
print(10 + 5 * 2)
```

Jika kita ingin penjumlahan dilakukan lebih dulu, kita harus menggunakan tanda kurung:
Sedangkan pada soal, 10 + 5 * 2 tidak ada tanda kurung. Hal ini menegaskan bahwa python mendahulukan prioritas yang lebih tinggi (perkalian (*)) untuk diselesaikan terlebih dahulu daripada penjumlahan (+)


```python
(10 + 5) * 2
print((10 + 5) * 2)
```

  2. Operator relasional digunakan untuk membandingkan dua nilai. Hasil dari perbandingan tersebut selalu berupa nilai Boolean, yaitu True atau False. Dokumentasi Python menyebutkan bahwa operator perbandingan (comparison operators) seperti >, <, ==, !=, >=, dan <= digunakan untuk membandingkan nilai.

```python
umur = 20
print(umur >=15)
print(umur < 15)
```

Sedangkan operator logika digunakan untuk menggabungkan atau memanipulasi beberapa kondisi Boolean. Operator logika utama dalam Python adalah and, or, dan not.

```python
print(True and False) # False
print(True or False) # True
print(not True) # False
```

Contoh dibawah ini menampilkan pyhton dengan dua jenis operator yang dipakai bersamaan dalam satu ekspresi untuk menghasilkan keputusan yang benar (True)

```python
nilai = 90
kehadiran = 100
diterima_beasiswa = (nilai >= 85) and (kehadiran >= 85)
print(diterima_beasiswa)
```

  3. Seorang kasir ingin menghitung total pembayaran dengan rumus:
(Harga + pajak ) * jumlah barang. Tapi dia menulis:


```python
harga = 100000
pajak = 10000
jumlah = 3
total = harga + pajak * jumlah

print(total)
```

Karena operator perkalian (*) memiliki prioritas lebih tinggi daripada penjumlahan (+), Python menghitung: 100000 + (10000 * 3) maka hasilnya Adalah 130000. Padahal yang diinginkan adalah (100000 + 100000) * 3 yang menghasilkan 330000. Maka perbaikan yang harus dilakukan:

```python
harga = 100000
pajak = 10000
jumlah = 3
total = (harga + pajak) * jumlah

print(total)
```
