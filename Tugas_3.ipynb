```python
# NAMA : Mailaf Faiza Yasmin
# NIM : 4232501039
# TANGGAL : 24 September 2026
# TUJUAN : Menampilkan hasil operasi aritmatika (+, -, *, /, //, %) dari dua angka

#b11 soal umum
bil1 = int (input("Masukkan bilangan pertama: "))
bil2 = int (input("Masukkan bilangan kedua: "))

print("\n--- Hasil operasi aritmatika ---")

hasil_tambah = bil1 + bil2
print(f"Hasil penjumlahan ({bil1} + {bil2})         : {hasil_tambah}")

hasil_kurang = bil1 - bil2
print(f"Hasil pengurangan ({bil1} - {bil2})         : {hasil_kurang}")

hasil_kali = bil1 * bil2
print(f"Hasil perkalian ({bil1} * {bil2})           : {hasil_kali}")

hasil_bagi = bil1 / bil2
print(f"Hasil pembagian ({bil1} / {bil2})           : {hasil_bagi}")

hasil_pembagian_bulat = bil1 // bil2
print(f"Hasil pembagian bulat ({bil1} // {bil2})    : {hasil_pembagian_bulat}")

hasil_moulus = bil1 % bil2
print(f"Hasil modulus ({bil1} % {bil2})             : {hasil_moulus}")
```
