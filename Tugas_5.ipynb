Soal 1 : Belanja Buku

Budi membeli 7 buku dengan harga Rp 18.500 per buku. Toko memberi diskon 10% dari total belanja. Budi membayar dengan uang Rp 150.000.

Tentukan:

1. Total belanja, besar diskon, dan jumlah yang harus dibayar.
2. Uang kembalian.
3. Jika jumlah yang dibayar dipatungi 4 orang secara adil dalam bilangan bulat, berapa iuran per orang dan berapa rupiah sisanya?
4. Apakah uang Budi cukup? (hasil True/False)

```python
#LATIHAN 5
#SOAL 1

jumlah_buku = 7
harga_per_buku = 18500
diskon_persen = 10
uang_bayar = 150000

total_belanja = jumlah_buku * harga_per_buku
diskon = total_belanja * diskon_persen / 100
jumlah_dibayar = total_belanja - diskon
kembalian = uang_bayar - jumlah_dibayar
iuran_per_orang = int(jumlah_dibayar // 4)
sisa = int(jumlah_dibayar % 4)
cukup = uang_bayar >= jumlah_dibayar

print("=== HASIL PERHITUNGAN ===")
print("Total Belanja (Rp) :", int(total_belanja))
print("Diskon (Rp) :", int(diskon))
print("Jumlah Harus Dibayar (Rp) :", int(jumlah_dibayar))
print("Uang Kembalian (Rp) :", int(kembalian))
print("Iuran per Orang (Rp) :", iuran_per_orang)
print("Sisa Pembagian (Rp) :", sisa)
print("Uang Budi Cukup? :", cukup)
```

Soal 2 : PLTS Atap Rumah

Sebuah rumah memasang 12 panel surya berdaya 450 Wp per panel. Rata-rata penyinaran efektif 5 jam/hari dengan efisiensi sistem 80% (0,8). Kebutuhan listrik rumah 25 kWh/hari dan tarif listrik Rp 1.444,70/kWh.

Tentukan:

1. Energi yang dihasilkan per hari (kWh).
2. Energi per bulan (30 hari) dan nilai rupiah listrik yang dihasilkan.
3. Kekurangan energi per hari terhadap kebutuhan.
Apakah PLTS sudah mencukupi kebutuhan rumah? (True/False)

```python
#LATIHAN 5
#SOAL 2

jumlah_panel = 12
daya_panel = 450 # Wp
jam_efektif_matahari = 5 # jam/hari
efisiensi = 0.8 # 80%
kebutuhan_rumah = 25 # kWh/hari
tarif_listrik = 1444.70 # Rp/kWh

energi_harian = (jumlah_panel * daya_panel * jam_efektif_matahari * efisiensi) / 1000 # kWh
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik
kekurangan_energi = kebutuhan_rumah - energi_harian
mencukupi = energi_harian >= kebutuhan_rumah

print("=== HASIL PERHITUNGAN PLTS ===")
print("Energi per Hari (kWh) :", energi_harian)
print("Energi per Bulan (kWh):", energi_bulanan)
print(f"Nilai Listrik/Bulan (Rp) : {nilai_listrik:.2f}")
print("Kekurangan Energi (kWh) :", round(kekurangan_energi, 2))
print("PLTS Mencukupi? :", mencukupi)
```

Soal 3 : Generator Diesel dan Rugi Kabel

Sebuah generator diesel 200 kW beroperasi 8 jam/hari. Konsumsi bahan bakar 0,25 liter per kWh, harga solar Rp 13.500/liter, dan kapasitas tangki 1.000 liter. Kabel penghantar dialiri arus 20 A dengan hambatan 0,5 Ω.

Tentukan:

1. Energi yang dibangkitkan per hari (kWh).
2. Kebutuhan solar per hari dan biayanya.
3. Berapa hari penuh tangki cukup, dan berapa liter solar yang tersisa?
4. Rugi daya pada kabel (P = I² × R).
5. Apakah rugi kabel tergolong tinggi, yaitu lebih dari 150 W dan arus minimal 20 A? (True/False)

```python
#LATIHAN 5
#SOAL 3

daya_generator = 200 # kW
jam_operasi = 8 # jam/hari
konsumsi_bahan_bakar = 0.25 # liter/kWh
harga_solar = 13500 # Rp/liter
kapasitas_tangki = 1000 # liter
arus = 20 # A
hambatan = 0.5 # ohm

energi_harian = daya_generator * jam_operasi
solar_harian = energi_harian * konsumsi_bahan_bakar
biaya_harian = solar_harian * harga_solar
hari_penuh = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian
rugi_daya = (arus ** 2) * hambatan
rugi_tinggi = (rugi_daya > 150) and (arus >= 20)

print("=== HASIL PERHITUNGAN ===")
print("Energi per Hari (kWh) :", energi_harian)
print("Solar per Hari (L) :", solar_harian)
print(f"Biaya Solar per Hari (Rp) : {biaya_harian:.1f}")
print("Tangki Cukup (hari) :", int(hari_penuh))
print("Sisa Solar (L) :", sisa_solar)
print("Rugi Daya Kabel (W) :", rugi_daya)
print("Rugi Kabel Tinggi? :", rugi_tinggi)
```
