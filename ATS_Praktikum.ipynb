2. Sebuah PLTD beroperasi dengan daya_kw = 450 selama jam = 8 jam. Dalam sehari (24 jam) PLTD tersebut menghasilkan energi_aktual = 7200 kWh dari kapasitas_kw = 400.


  a. Simpan nilai dalam variabel, hitung energi harian (energi = daya_kw × jam), lalu tampilkan dengan f-string.
  
  b. Hitung capacity factor: Capacity factor (%) = (energi_aktual / (kapasitas_kw × 24)) × 100. Tampilkan dengan 2 angka desimal.

  Contoh output:

  Energi harian = 3600 kWh
  Capacity factor = 75.00%

```python
#SOAL 2: ENERGI HARIAN DAN CAPACITY FACTOR PLTD

daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

energi = daya_kw * jam

capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

print(f"Energi harian = {energi} kWh")
print(f"Capacity factor = {capacity_factor:.2f}%")
```

6. Status sebuah boiler ditentukan dari suhu uap (°C) dan tekanan (bar):

"STATUS: TRIP" jika suhu mencapai 550 atau lebih, atau tekanan mencapai 14 atau lebih.
"STATUS: WARNING" jika suhu lebih dari 520 dan tekanan lebih dari 12.
selain itu, "STATUS: NORMAL".
Baca suhu dan tekanan dari input (bertipe float). Gunakan if-elif-else dengan operator logika. Perhatikan urutan pemeriksaan kondisi.

Uji program dengan kombinasi berikut:

image.png

```python
#SOAL 6: STATUS OPERASI BOILER

suhu = float(input("Masukkan suhu uap (°C): "))
tekanan = float(input("Masukkan tekanan (bar): "))

if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")
elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")
else:
    print("STATUS: NORMAL")
```

8. Simulasikan pengosongan tangki dari 100 liter hingga kosong. Mulai dengan volume = 100. Gunakan perulangan while: selama volume > 0, kurangi 20 liter, lalu cetak volume saat ini (setelah dikurangi). Saat volume mencapai tepat 20 liter, cetak "WARNING: Hampir kosong!". Setelah loop selesai, cetak pesan tangki kosong.

Contoh output:

Volume: 80 liter

Volume: 60 liter

Volume: 40 liter

Volume: 20 liter
WARNING: Hampir kosong!

Volume: 0 liter
TANGKI KOSONG - Pengosongan berhenti.

```python
#SOAL 8: PENGOSONGAN TANGKI BAHAN BAKAR

volume = 100

while volume > 0:
    volume -= 20
    print(f"Volume: {volume} liter")

    if volume == 20:
        print("WARNING: Hampir kosong!")
        print("TANGKI KOSONG - Pengosongan berhenti.")
```

10. Data suhu bearing lima titik pengukuran dibaca sebagai teks, dan satu sensor rusak mengirim "N/A":

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

Periksa setiap data dengan for. Nomor titik dimulai dari 1.

a. Ubah data menjadi float dengan try-except ValueError. Jika gagal, cetak Titik [n]: data tidak valid - pemeriksaan dihentikan, lalu hentikan perulangan dengan break.

b. Jika nilai melebihi 70.0, cetak Warning: Titik [n] suhu bearing tinggi!.

c. Hitung jumlah data valid yang berhasil diperiksa, dan tampilkan setelah perulangan selesai.

Output yang diharapkan untuk data di atas:

Warning: Titik 3 suhu bearing tinggi!

Titik 4: data tidak valid - pemeriksaan dihentikan

Jumlah data valid: 3

```python
#SOAL 10: PEMBACAAN SUHU BEARING DENGAN SENSOR RUSAK

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

jumlah_valid = 0

for i, nilai in enumerate(data, start=1):
    try:
        suhu = float(nilai)
        jumlah_valid += 1
        if suhu > 70.0:
            print(f"Warning: Titik {i} suhu bearing tinggi!")

    except ValueError:
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break

print(f"Jumlah data valid: {jumlah_valid}")
```

Buatlah program yang menyelesaikan satu masalah di bidang energi (mis. pemantauan beban harian, deteksi anomali suhu, simulasi pengisian/pengosongan, atau penghitungan konsumsi listrik). Program minimal memuat:

- percabangan if-elif-else,

Tambahkan komentar yang menjelaskan masalah dan cara kerja program. Penggunaan try-except dan variabel/ekspresi yang tepat akan dinilai lebih baik.

```python
#CHALLENGE

beban = [120, 150, 180, 210, 250]

print("=== PEMANTAUAN BEBAN LISTRIK ===")

for i, nilai in enumerate(beban, start=1):
    try:
        daya = float(nilai)
        print(f"\nData ke-{i}: {daya} kW")

        if daya >= 250:
            print("Status: BEBAN TINGGI")
        elif daya >= 180:
            print("Status: BEBAN SEDANG")
        else:
            print("Status: BEBAN NORMAL")

    except ValueError:
        print(f"Data ke-{i} tidak valid!")

print("\nPemantauan beban listrik selesai.")
```
