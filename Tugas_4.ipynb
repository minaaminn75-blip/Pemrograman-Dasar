```python
#LATIHAN 4
# 1. MENENTUKAN TAHUN KABISAT
tahun = int(input("Masukkan tahun: "))
is_kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

if is_kabisat:
    print(f"Tahun {tahun} adalah tahun kabisat.")
else:
    print(f"Tahun {tahun} BUKAN tahun kabisat.")
```

```python
# 2. MEMBUAT PROGRAM SEDERHANA ARITMATIKA
nama_barang = input("Masukkan nama barang: ")
harga_satuan = float(input("Masukkan harga per unit (Rp): "))
jumlah_beli = int(input("Masukkan jumlah yang dibeli: "))
persen_diskon = float(input("Masukkan persen diskon (%): "))

total_awal = harga_satuan * jumlah_beli

nilai_diskon = total_awal * (persen_diskon / 100)

total_bayar = total_awal - nilai_diskon

print("\n" + "=" * 35)
print("       Rincian Pembayaran")
print("=" * 35)
print(f"Barang        : {nama_barang}")
print(f"Harga Satuan  : Rp {harga_satuan:,.2f}")
print(f"Jumlah Beli   : {jumlah_beli}")
print(f"Total Awal    : Rp {total_awal:,.2f}")
print(f"Diskon ({persen_diskon}%) : Rp {nilai_diskon:,.2f}")
print("-" * 35)
print(f"Total Bayar   : Rp {total_bayar:,.2f}")
print("=" * 35)
```
