```python
#LATIHAN 4
# 1. MENENTUKAN TAHUN KABISAT

tahun = int(input("Masukkan tahun: "))
kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

if kabisat:
    print(f"Tahun {tahun} adalah tahun kabisat.")
else:
    print(f"Tahun {tahun} BUKAN tahun kabisat.")
```

```python
# 2. MEMBUAT PROGRAM SEDERHANA DENGAN LINGKUP FUNGSI INPUT, FUNGSI OUTPUT DAN OPERASI

panjang = float(input("Masukkan panjang balok: "))
lebar = float(input("Masukkan lebar balok: "))
tinggi = float(input("Masukkan tinggi balok: "))

volume = panjang * lebar * tinggi

print("HASIL PERHITUNGAN VOLUME BALOK")
print("panjang :", panjang)
print("lebar :", lebar)
print("tinggi :", tinggi)
print("volume :", volume)
```
