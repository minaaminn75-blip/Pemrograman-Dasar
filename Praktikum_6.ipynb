```python
# IF - ELIF - ELSE: PERCABANGAN MAJEMUK

nilai = int(input("NILAI ULANGAN HARIAN: "))
if nilai >= 85:
    print("Wowww amazing! Nilai: A")
elif nilai >= 70:
    print("Good jobb! Nilai: B")
elif nilai >= 60:
    print("Very well! Nilai: C")
else:
    print("Nilai kurang, belajar lagi ya!! :)")
```

```python
# MENENTUKAN JENIS DISKON BERDASARKAN HARI

hari = input("HARI BELANJA: ")

if hari == "Sabtu" or hari == "Minggu":
    diskon: float = 0.50 # diskon 50%
    print(f"YEYY! SELAMAT YA KAMU DAPAT DISKON WEEKEND SEBESAR {diskon*100}%.")
elif hari == "Jumat":
    diskon: float = 0.20 # diskon 20%
    print(f"DISKON SPESIAL JUMAT BERKAH!! SELAMAT YA KAMU DAPAT POTONGAN BELANJA {diskon*100}%.")
else:
    diskon: float = 0.10 # diskon 10%
    print(f"KAMU DAPAT DISKON WEEKDAY SEBESAR {diskon*100}%.")
```

```python
# KONVERSI MATA UANG

rupiah = int(input("Masukkan jumlah uang (Rp): "))

print("1. Dollar (USD)")
print("2. Euro (EUR)")
print("3. Yen (JPY)")

pilih = int(input("Pilih konversi (1/2/3): "))

if pilih == 1:
    print("Dalam Dollar:", rupiah / 15000)
elif pilih == 2:
    print("Dalam Euro:", rupiah / 16000)
elif pilih == 3:
    print("Dalam Yen:", rupiah / 120)
else:
    print("Pilihan tidak valid")
```
